```python
from google_play_scraper import app
import pandas as pd
import numpy as np
```

```python
#Scrape desired number of reviews
from google_play_scraper import Sort, reviews

result, continuation_token = reviews(
    'id.toco',
    lang='id', # defaults to 'en'
    country='id', # defaults to 'us'
    sort=Sort.MOST_RELEVANT, # defaults to Sort.MOST_RELEVANT you can use Sort.NEWEST to get newst reviews
    count=2000, # defaults to 100
    filter_score_with=None # defaults to None(means all score) Use 1 or 2 or 3 or 4 or 5 to select certain score
)
```

```python
toco = pd.DataFrame(np.array(result),columns=['content'])
toco = toco.join(pd.DataFrame(toco.pop('content').tolist()))
toco.head(11)
```

```python
toco_df = toco[['at','content','score']]
toco_df = toco_df.sort_values(by='at', ascending=False) #Sort by Newst, change to True if you want to sort by Oldest.
toco_df.head(20)
```

```python
toco_df.to_csv("scrap_toco3.csv", index = False)  #Simpan hasil data scraping ke csv

from google.colab import files
files.download('scrap_toco3.csv')  # Unduh data hasil scraping
```

2. Labeling

```python
label = []
for index, row in toco_df.iterrows():
  if row['score'] == 1 or row['score'] == 2:
    label.append(0)
  else :
    label.append(1)
toco_df["label"] = label
```

```python
toco_df.head(20)
```

3. Preprocessing

```python
#casefolding
toco_df['content'] = toco_df['content'].str.lower()
toco_df.head(20)
```

```python
#Normalisasi
norm = {"bgt" : "banget", "yg" :"yang", "sdh" : "sudah", "apk" : "aplikasi", "utk" : "untuk", "sma" : "sama", "app" : "aplikasi"}

def normalisasi(str_text):
  for i in norm:
    str_text = str_text.replace(i, norm[i])
  return str_text

toco_df['content'] = toco_df['content'].apply(lambda x: normalisasi(x))
toco_df.head(100)
```

```python
import numpy as np
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(decode_error='replace', encoding='utf-8')
```

```python
x_train = vectorizer.fit_transform(x_train_raw)
x_test = vectorizer.transform(x_test_raw)

print(x_train.shape)
print(x_test.shape)
```

```python
x_train = x_train.toarray()
```

```python
x_test = x_test.toarray()
```
